Portal akademik fiktif memiliki alur yang dianggap membingungkan. Kita menganalisis success rate dan task time pada data uji sintetis, lalu menyusun prioritas perbaikan. Data ini bukan hasil pengujian pengguna nyata dan bukan audit WCAG formal.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

persona = pd.DataFrame([
    {"persona":"Nadia — Mahasiswa baru", "goals":"Menemukan jadwal dan mengunduh surat",
     "pain_points":"Istilah menu tidak konsisten; takut salah mengirim formulir",
     "behaviors":"Lebih sering memakai ponsel; perlu instruksi jelas"},
    {"persona":"Bima — Asisten akademik", "goals":"Memeriksa status permohonan mahasiswa",
     "pain_points":"Harus membuka banyak halaman; status tidak terlihat",
     "behaviors":"Sering mengulang tugas; butuh pencarian cepat"}
])
display(persona)
```

```python
requirements = pd.DataFrame([
    {"id":"F-01","jenis":"Fungsional","kebutuhan":"Pengguna dapat mencari jadwal berdasarkan semester"},
    {"id":"F-02","jenis":"Fungsional","kebutuhan":"Pengguna dapat melihat status permohonan"},
    {"id":"NF-01","jenis":"Non-fungsional","kebutuhan":"Navigasi konsisten dan dapat digunakan dengan keyboard"},
    {"id":"NF-02","jenis":"Non-fungsional","kebutuhan":"Pesan kesalahan menjelaskan cara memperbaiki input"},
    {"id":"NF-03","jenis":"Non-fungsional","kebutuhan":"Kontras dan ukuran teks mendukung keterbacaan"}
])
display(requirements)
```

```python
# Data uji simulasi: setiap baris satu peserta-tugas. Bukan hasil riset nyata.
test = pd.DataFrame([
    ("Jadwal", "P01", 1, 42), ("Jadwal", "P02", 1, 51), ("Jadwal", "P03", 1, 39),
    ("Surat", "P01", 0, 120), ("Surat", "P02", 1, 98), ("Surat", "P03", 0, 120),
    ("Status permohonan", "P01", 0, 120), ("Status permohonan", "P02", 1, 86), ("Status permohonan", "P03", 0, 120),
    ("Ubah profil", "P01", 1, 35), ("Ubah profil", "P02", 1, 32), ("Ubah profil", "P03", 1, 41),
], columns=["tugas","peserta","berhasil","waktu_detik"])
summary = test.groupby("tugas").agg(
    jumlah_percobaan=("berhasil","size"),
    jumlah_berhasil=("berhasil","sum"),
    success_rate=("berhasil","mean"),
    rata_waktu_detik=("waktu_detik","mean")
).reset_index()
summary["success_rate_pct"] = (summary["success_rate"]*100).round(1)
display(summary.sort_values("success_rate"))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
s = summary.sort_values("success_rate")
axes[0].barh(s["tugas"], s["success_rate_pct"])
axes[0].set_xlim(0, 100)
axes[0].set_xlabel("Success rate (%)")
axes[0].set_title("Keberhasilan per tugas")
axes[1].barh(s["tugas"], s["rata_waktu_detik"])
axes[1].set_xlabel("Rata-rata detik")
axes[1].set_title("Waktu tugas rata-rata")
plt.tight_layout()
plt.show()
```

```python
heuristics = pd.DataFrame([
    {"temuan":"Status permohonan sulit ditemukan","prinsip":"Visibility of system status","dampak":5,"usaha_perbaikan":2},
    {"temuan":"Label menu berbeda untuk fungsi serupa","prinsip":"Consistency and standards","dampak":4,"usaha_perbaikan":2},
    {"temuan":"Pesan kesalahan tidak menyebut langkah berikutnya","prinsip":"Help users recognize/recover from errors","dampak":4,"usaha_perbaikan":1},
    {"temuan":"Tombol utama kecil di layar ponsel","prinsip":"Operable / affordance","dampak":3,"usaha_perbaikan":2},
])
heuristics["rasio_prioritas"] = heuristics["dampak"] / heuristics["usaha_perbaikan"]
display(heuristics.sort_values("rasio_prioritas", ascending=False))
```
